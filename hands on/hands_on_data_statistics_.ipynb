```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/StudentsPerformance.csv")

print("Student Dataset")
display(df)

def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["math score"].apply(grade)

frequency = df["Grade"].value_counts().sort_index()

print("\nFrequency Distribution of Grades")
print(frequency)

plt.figure(figsize=(6,5))
plt.bar(frequency.index, frequency.values)
plt.title("Frequency Distribution of Students' Grades")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)
df = housing.frame

print("House Price Dataset")
display(df.head())

df["House Price"] = housing.target * 100000

plt.figure(figsize=(8,5))
plt.hist(df["House Price"], bins=20, edgecolor="black")
plt.title("Histogram of House Prices")
plt.xlabel("House Price")
plt.ylabel("Frequency")
plt.show()

skew = df["House Price"].skew()

print("Skewness:", round(skew, 2))

if skew > 0:
    print("The data is Positively Skewed.")
elif skew < 0:
    print("The data is Negatively Skewed.")
else:
    print("The data is Normally Distributed.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("/content/core_dataset.xlsx")

print("Employee Dataset")
display(df.head())

plt.figure(figsize=(6,5))
plt.boxplot(df["Pay Rate"].dropna()) # Changed "PayRate" to "Pay Rate" to match the column name from `qtrMJAqOFhik`
plt.title("Box Plot of Employee Salary")
plt.ylabel("Salary")
plt.show()

Q1 = df["Pay Rate"].quantile(0.25)
Q3 = df["Pay Rate"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Pay Rate"] < lower) | (df["Pay Rate"] > upper)]

print("Outliers")
display(outliers[["Employee Name", "Pay Rate"]]) # Changed "Employee_Name" to "Employee Name" to match the column name

print("\nNumber of Outliers:", len(outliers))
```

```python
import pandas as pd

df = pd.read_excel("/content/core_dataset.xlsx")

print("Original Dataset")
display(df.head())

Q1 = df["Pay Rate"].quantile(0.25)
Q3 = df["Pay Rate"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

cleaned_df = df[(df["Pay Rate"] >= lower) & (df["Pay Rate"] <= upper)]

print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", cleaned_df.shape)

print("\nCleaned Dataset")
display(cleaned_df)

cleaned_df.to_csv("Cleaned_HRDataset.csv", index=False)

print("\nCleaned dataset saved as 'Cleaned_HRDataset.csv'")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/Iris.csv")

print("Iris Dataset")
display(df)

print("\nDataset Information")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nDescriptive Statistics")
print(df.describe())

df.hist(figsize=(10,8), edgecolor="black")
plt.show()

numeric_columns = ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]

for column in numeric_columns:
    plt.figure(figsize=(5,5))
    plt.boxplot(df[column].dropna())
    plt.title("Box Plot of " + column)
    plt.ylabel(column)
    plt.show()

cleaned_df = df.copy()

for column in numeric_columns:
    Q1 = cleaned_df[column].quantile(0.25)
    Q3 = cleaned_df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = cleaned_df[(cleaned_df[column] < lower) | (cleaned_df[column] > upper)]

    print("\nOutliers in", column)
    display(outliers)

    cleaned_df = cleaned_df[
        (cleaned_df[column] >= lower) &
        (cleaned_df[column] <= upper)
    ]

print("\nCleaned Dataset")
display(cleaned_df)

cleaned_df.to_csv("Cleaned_Iris.csv", index=False)

print("\nCleaned dataset saved as Cleaned_Iris.csv")
```
